## “Iris Flower Clustering Project” 
Dataset:
Iris Dataset (UCI / Kaggle)
Tasks:
● Apply K-Means (k=3)
● Visualize clusters (scatter plot)
● Compare predicted clusters vs true labels
Skill Gain: Unsupervised ML + Visualization

```python
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target
```

```python
print("Dataset shape:", X.shape)
print("Feature names:", iris.feature_names)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans 

kmeans = KMeans(n_clusters = 3,random_state = 42, n_init = 10)
predited_clusters = kmeans.fit_predict(X_scaled)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
```

```python
plt.subplot(1, 2, 1)
plt.scatter(
    X[:, 2], X[:, 3],
    c=predicted_clusters,
    cmap="viridis",
    s=50
)
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("K-Means Predicted Clusters")
```

```python
plt.subplot(1, 2, 2)
plt.scatter(
    X[:, 2], X[:, 3],
    c=y,
    cmap="viridis",
    s=50
)
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.title("Actual Iris Species")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

comparison = pd.crosstab(
    pd.Series(y, name="Actual Species"),
    pd.Series(predicted_clusters, name="Predicted Cluster")
)

print("\nActual Species vs Predicted Clusters:")
print(comparison)
```

```python
from sklearn.metrics import adjusted_rand_score

ari = adjusted_rand_score(y, predicted_clusters)
print("\nAdjusted Rand Index:", round(ari, 3))
```
