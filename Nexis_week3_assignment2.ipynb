## Apply PCA to reduce dataset dimensions.

PCA is a dimensionality reduction technique used in machine learning to reduce the number of features in a dataset while preserving as much information (variance) as possible.

Original dataset
4 features
Sepal length · Sepal width · Petal length · Petal width

Apply PCA

Reduced dataset
2 principal components
PC1 · PC2

```python
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data

print("Original dataset shape:", X.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

PCA is affected by feature scales. Standardization ensures that features with larger numerical ranges do not automatically dominate the principal components.

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance preserved:", pca.explained_variance_ratio_.sum())
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=iris.target,
    cmap="viridis",
    s=50
)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA on Iris Dataset")
plt.colorbar(label="Flower Species")
plt.show()
```
